```python
#list in python
#list is an ordered and changeable collection can store multiple values
marks = [80, 90, 75, 85]

print(marks)
```

```python
#accessing elements in a list
marks = [80, 90, 75, 85]

print(marks[0])
print(marks[1])
print(marks[2])
```

```python
#changing elements in a list
marks = [80, 90, 75]

marks[1] = 95

print(marks)
```

```python
#change elements in a list
marks = [79, 64, 92, 74,96, 77, 94]

marks[1] = 56
marks[3] = 89
marks[5] = 59

print(marks)
```

```python
#add elements to a list
marks = [80, 90, 75]

marks.append(85)

print(marks)
```

```python
#add elements to a list
marks = [48, 90, 78]

marks.append(89)

print(marks)
```
